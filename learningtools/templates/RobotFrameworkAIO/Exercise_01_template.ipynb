## Practice 2: Test suite with a custom configuration file

In this practice you define a configuration file in JSONP format and pass it to the robot test suite.

The `json_config` variable holds the JSONP configuration content. It is written to `config_file_path` before the robot suite runs.

In the `*** Variables ***` in the `robot_code` section is filled with the actual path at runtime:

```robot
*** Variables ***
${config_file}   <path interpolated from config_file_path>
```

The `testsuite_setup` keyword reads `${config_file}` to load the configuration. The config parameters in side `params['global']` will be accessed as global variables in the robot test script.

````python
# Create a JSON configuration file for the test suite.
json_config = """
{
    "WelcomeString" : "Hello... Robot Framework is running now!",
    "Maximum_version" : "1.0.0",
    "Minimum_version" : "0.6.0",
    "Project" : "RobotFramework Testsuites",
    "TargetName" : "Practice-02",
    "params": {
        // global parameters
        "global" : {
            "robot_variable_1" : "ABC",
            "robot_variable_2" : 25
        }
    }
}
"""
# Write the JSON configuration to a file.
config_file_path = os.path.abspath("./learningtools/examples/RobotFrameworkAIO/practice2_config.jsonp").replace("\\", "/")
with open(config_file_path, "w") as temp_file:
    temp_file.write(json_config)

# Create a Robot Framework test suite that uses the configuration file.
robot_code = f"""
*** Variables ***
${{<your input here>}}   {config_file_path}

*** Settings ***
Library    RobotFramework_TestsuitesManagement    WITH NAME    tm
Suite Setup    tm.testsuite_setup

*** Test Cases ***
Test Case practice-02
    [documentation]    practice-02
    Should Be Equal    ${{<your input here>}}    ABC
    Should Be Equal    ${{<your input here>}}    ${{25}}
"""

p2.check()

os.remove(config_file_path)
````
