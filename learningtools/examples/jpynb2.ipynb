That mirrors the Kaggle Learn pattern where the lesson shell stays stable while the exercise content changes per notebook.

```python
from pathlib import Path
import sys

def _resolve_notebook_path(default_name: str) -> Path:
    try:
        return Path(__vsc_ipynb_file__).resolve()
    except NameError:
        pass

    cwd = Path.cwd()
    candidates = [
        cwd / default_name,
        cwd / "learningtools" / "examples" / default_name,
        cwd.parent / "examples" / default_name,
    ]
    for candidate in candidates:
        if candidate.is_file():
            return candidate.resolve()
    return (cwd / default_name).resolve()

notebook_path = _resolve_notebook_path("jpynb2.ipynb")
LEARNINGTOOLS_NOTEBOOK = str(notebook_path)

repo_root = next(
    parent
    for parent in notebook_path.parents
    if (parent / "learningtools" / "__init__.py").is_file()
    and (parent / "learningtools" / "core.py").is_file()
    and (parent / "learningtools" / "examples").is_dir()
    and (parent / "learningtools" / "workbooks").is_dir()
)
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from learningtools import *

lesson
```
